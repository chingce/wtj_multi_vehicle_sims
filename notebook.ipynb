```python
from sim import sim
import matplotlib.pyplot as plt
```

```python
n_vehicles = 100

data, time, platoon = sim(n_vehicles=n_vehicles, penetration_rate=0.2, sim_length_seconds=360)
```

```python
for i in range(data.shape[0]):
    if i == 0:
        plt.plot(time, data[i, 0, :], color='green')
    elif platoon[i-1].type == 'WTJ':
        plt.plot(time, data[i, 0, :], color='blue')
    else:
        plt.plot(time, data[i, 0, :], color='red')

# plt.ylim(-50*n_vehicles, 1200)
plt.ylim(-2000, 0)
plt.xlim(0, 200)
plt.show()
```

```python
import numpy as np
from sim import dynamics
time = np.arange(0, 6, 0.05)
cmd_accel = np.zeros(120)
cmd_accel[20:] = 3

accel = np.zeros_like(cmd_accel)
accel[0] = 0  # Initial acceleration
for i in range(1, len(time)):
    accel[i] = dynamics(cmd_accel[i], accel[i-1])

plt.plot(time, cmd_accel)
plt.plot(time, accel)
```
